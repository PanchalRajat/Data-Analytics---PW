###Question 1: What is the key difference between publishing a dashboard on Tableau Public versus Tableau Server/Cloud in terms of accessibility and data security?

-->

Answer:

This question explains the difference between Tableau Public and Tableau Server/Cloud for dashboard sharing and security.

Steps:

Publish dashboard on Tableau Public
Dashboard becomes publicly accessible on the internet
Anyone with the link can view the dashboard
Data security and user restrictions are limited
Publish dashboard on Tableau Server or Tableau Cloud
Set user authentication and access permissions
Allow only authorized users to access the dashboard
Protect sensitive business data using security controls

Insight:
Tableau Public is open for everyone, while Tableau Server/Cloud provides controlled and secure access.

Business Value:
Helps organizations securely share dashboards while protecting confidential business data.

###Question 2: When publishing a Tableau dashboard, what happens to a live connection compared to an extract? How does this affect data refresh and performance?

-->

Answer:

This question explains how live connections and extracts behave after publishing a dashboard.

Steps:

Use Live Connection to connect directly with the original database
Published dashboard always shows latest data from the source
Dashboard performance depends on database speed and network
Use Extract to create a snapshot copy of data
Publish the extract with the dashboard
Schedule extract refresh if updated data is needed
Extracts usually load faster than live connections
Extracts reduce load on the main database server

Insight:
Live connections provide real-time data, while extracts improve dashboard performance.

Business Value:
Helps organizations balance between real-time reporting and faster dashboard speed.

###Question 4: If you want users to view a dashboard but not download the underlying data, which permission setting would you adjust in Tableau Server?

-->

Answer:

This question explains how to restrict users from downloading underlying data.

Steps:

Open Tableau Server or Tableau Cloud
Go to the published dashboard settings
Open Permissions
Select the user or group
Find the Download Data or View Underlying Data permission
Set the permission to Deny
Save the permission changes
Users can now view dashboards without downloading data

Insight:
Restricting download permissions protects sensitive business information.

Business Value:
Helps organizations maintain data privacy and prevent unauthorized data sharing.

###Question 6: List two best practices you should follow when preparing a dashboard for public sharing to ensure data security and privacy.

-->

Answer:

This question explains important practices for protecting data before public sharing.

Steps:

Remove confidential or personal information from the dataset
Hide sensitive fields like passwords, phone numbers, or customer IDs
Use aggregated data instead of detailed records when possible
Check all filters and hidden sheets before publishing
Verify user permissions and sharing settings
Test the dashboard using a public access view
Ensure only safe and required data is visible
Publish the dashboard after final security verification

Insight:
Careful data review helps prevent accidental exposure of sensitive information.

Business Value:
Protects company privacy, customer data, and organizational reputation.

###Question 7: After publishing a dashboard, users report a “Data not found” error. What is the most common cause related to data source settings?

-->

Answer:

This question explains the common reason behind “Data not found” errors after publishing.

Steps:

Open the published dashboard settings
Check the data source connection
Verify database server name and credentials
Confirm the data source path is correct
Check whether the database is accessible from Tableau Server or Cloud
Verify permissions for the database account
Refresh or republish the data source if required
Test the dashboard connection again

Insight:
Incorrect or broken data source connections are the most common cause of this error.

Business Value:
Helps maintain dashboard reliability and prevents reporting interruptions.
