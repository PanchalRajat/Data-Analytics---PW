### Question 1: What is Data Loading in Power BI and why is it considered the first step of analysis?

-->  

Answer:  
Data Loading in Power BI refers to the process of importing data from different sources (such as Excel, CSV files, databases, or cloud services) into Power BI so that it can be used for analysis and visualization.

It is considered the **first step of analysis** because no analysis can be performed without data. Before cleaning, transforming, or visualizing data, it must first be loaded into the Power BI environment.

During data loading, Power BI establishes a connection with the data source and brings the data into the tool. This step ensures that all required data is available for further processing such as data cleaning, transformation, modelling, and reporting.

Therefore, data loading is the foundation of the entire data analysis process in Power BI.

### Question 2: Explain the difference between “Load” and “Transform Data” in Power BI.

-->  

Answer:  
In Power BI, both **Load** and **Transform Data** are options used when importing data, but they serve different purposes.

**Load** directly imports the data into Power BI without making any changes. It is used when the dataset is already clean and ready for analysis. This option quickly loads the data into the model so that reports and visualizations can be created immediately.

**Transform Data**, on the other hand, opens the Power Query Editor, where users can clean and modify the data before loading it. This includes tasks such as removing duplicates, handling missing values, changing data types, filtering rows, and splitting columns.

In simple terms, **Load is used when data is already prepared**, while **Transform Data is used when data needs cleaning and transformation before analysis**.

### Question 3: What is a Fact Table and a Dimension Table? Give examples from the dataset.

-->  

Answer:  
In data modelling, tables are generally divided into **Fact Tables** and **Dimension Tables**.

A **Fact Table** contains quantitative data (measurable values) and is used for analysis. It usually includes numerical values and foreign keys that link to dimension tables.  
In this dataset, columns like **Confirmed, Deaths, Recovered** represent facts because they are numeric values used for analysis.

A **Dimension Table** contains descriptive or categorical data that provides context to the facts. It includes attributes such as names, categories, or dates.  
In this dataset, columns like **Country/Region** and **Date** act as dimensions because they describe when and where the data was recorded.

For example:  
- Fact data → Confirmed cases, Deaths, Recovered  
- Dimension data → Country/Region, Date  

Fact tables answer questions like “how many?”, while dimension tables answer “where?” and “when?”.

### Question 4: Why is Star Schema preferred over Snowflake Schema in Power BI?

-->  

Answer:  
Star Schema is preferred over Snowflake Schema in Power BI because it is simpler, faster, and more efficient for analysis and reporting.

In a **Star Schema**, a central fact table is directly connected to dimension tables. This structure is easy to understand and reduces the number of joins required during queries, which improves performance.

In contrast, a **Snowflake Schema** has normalized dimension tables that are split into multiple related tables. This increases complexity and requires more joins, which can slow down query performance in Power BI.

Star Schema is also easier to maintain and works well with Power BI’s in-memory engine, making data modeling and visualization faster and more user-friendly.

Therefore, Star Schema is preferred because it provides **better performance, simpler design, and easier analysis** compared to Snowflake Schema.

### Question 5: Identify and remove duplicate records based on Date, Country, and State.

-->  

Answer:  
To identify and remove duplicate records in Power BI, the combination of **Date, Country, and State** should be used as a unique key.

Steps to perform this in Power Query:

1. Click on **Transform Data** to open Power Query Editor.  
2. Select the columns **Date, Country, and State** (hold Ctrl and select all three).  
3. Go to the **Home tab → Remove Rows → Remove Duplicates**.  
4. Power BI will keep the first occurrence and remove duplicate rows based on the selected columns.  
5. Click **Close & Apply** to save the changes.

This ensures that only unique records remain in the dataset, preventing duplication in analysis and reports.

### Question 7: Create a new column to calculate Recovery Rate.

-->  

Answer:  
Recovery Rate is calculated as the percentage of recovered cases out of total confirmed cases.

Formula:  
Recovery Rate = (Recovered / Confirmed) × 100  

Steps to create this in Power BI:

1. Go to **Data View** or **Report View**.  
2. Click on **New Column** (or New Measure if needed).  
3. Enter the following DAX formula:

```DAX
Recovery Rate = ( [Recovered] / [Confirmed] ) * 100
```

4. Press **Enter**.  
5. Format the column as **Percentage** if required.

This new column helps analyze how many patients have recovered compared to total confirmed cases, which is useful for tracking recovery trends.

### Question 8: Create a summarized table showing total confirmed cases by Country.

-->  

Answer:  
To create a summarized table showing total confirmed cases by country in Power BI, aggregation can be done using DAX or Power BI visuals.

Using DAX:

```DAX
Country_Summary =
SUMMARIZE(
    coronavirus_2020_dataset,
    coronavirus_2020_dataset[Country],
    "Total_Confirmed", SUM(coronavirus_2020_dataset[Confirmed])
)
```

Alternatively, using Power BI interface:

1. Go to **Report View**.  
2. Add a **Table visual**.  
3. Drag **Country** into Rows.  
4. Drag **Confirmed** into Values (set aggregation to Sum).  

This will display total confirmed cases for each country, helping in country-wise comparison of COVID-19 impact.
