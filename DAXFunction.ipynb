### Question 1: Create a calculated column TotalPrice using Quantity and unit price
-->
Answer:

1. Go to **Data View** in Power BI
2. Select the table `amazon_store_sales_2022`
3. Click on **New Column**
4. Enter the formula:
   TotalPrice = amazon_store_sales_2022[Quantity] * amazon_store_sales_2022[UnitPrice]
5. Press **Enter** to create the column

This creates total price by multiplying quantity with unit price for each row.

### Question 2: Create a calculated column DiscountedPrice
-->

### Question 3: Create a measure average order value (AOV)
-->
Answer:

1. Go to **Report View** in Power BI
2. Select the table `amazon_store_sales_2022`
3. Click on **New Measure**
4. Enter the formula:
   AOV = DIVIDE(SUM(amazon_store_sales_2022[DiscountedPrice]), DISTINCTCOUNT(amazon_store_sales_2022[OrderID]))
5. Press **Enter** to create the measure

This calculates average revenue per order using total sales divided by number of orders.

### Question 4: Using DAX, create a measure that calculates cumulative Year-to-Date (YTD) Sales based on the Order Date
-->
Answer:

2. Select the table `amazon_store_sales_2022`

### Question 5: Using DAX, create a measure to rank each product according to its Total Sales, with the highest-selling product ranked first.
-->
Answer:

1. Go to **Report View**
2. Select table `amazon_store_sales_2022`
3. Click **New Measure**
4. Enter the formula:
   Product Rank = RANKX(ALL(amazon_store_sales_2022[Product]), SUM(amazon_store_sales_2022[DiscountedPrice]), , DESC, DENSE)
5. Press **Enter**

This ranks products based on total sales (highest = rank 1).

### Question 8: Create a measure called "Date Difference" that calculates the number of days between the latest Order Date and today's date using the DATEDIFF function.
-->
Answer:

1. Go to **Report View**
2. Select table `amazon_store_sales_2022`
3. Click **New Measure**
4. Enter the formula:
   Date Difference = DATEDIFF(MAX(amazon_store_sales_2022[Order Date]), TODAY(), DAY)
5. Press **Enter**

This calculates number of days from latest order date to today.
