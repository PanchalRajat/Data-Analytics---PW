###Question 1: Explain the difference between a Tableau Dashboard and a Tableau Story Point. Give two healthcare-related use cases.

-->

Answer:

This question explains the difference between Tableau Dashboards and Story Points with healthcare examples.

Steps:

Understand that a Tableau Dashboard combines multiple charts and visualizations into one interactive screen
Use dashboards to monitor data in real time using filters and actions
Understand that a Tableau Story Point presents dashboards or worksheets in a step-by-step sequence
Use Story Points to explain insights and create data storytelling
In healthcare, use a Dashboard to monitor patient admissions, treatment costs, and department performance together
Use another Dashboard to track patient outcomes, average stay duration, and regional hospital activity
Use a Story Point to explain how patient admissions changed across regions over time
Use another Story Point to present treatment cost analysis followed by department-wise performance insights

Insight:
Dashboards are used for interactive monitoring, while Story Points are used for presenting insights in a guided flow.

Business Value:
Helps hospital management both monitor healthcare operations and communicate insights effectively for decision-making.

###Question 2: What are Actions in Tableau? Explain how Filter Action can help hospital management analyze patient data.

-->

Answer:

This question explains Tableau Actions and how Filter Actions improve healthcare data analysis.

Steps:

Understand that Actions in Tableau create interaction between worksheets and dashboards
Common Tableau Actions include Filter Action, Highlight Action, and URL Action
Create multiple worksheets such as Department Analysis and Patient Outcome Analysis
Open Dashboard and add all worksheets
Click Dashboard → Actions
Select Add Action → Filter
Choose Source Sheet and Target Sheet
Set Run Action On as Select
Choose fields like Department or Hospital_Region for filtering
Click OK to apply the Filter Action
Select a department in one chart to filter related patient data in other charts
Analyze patient count, treatment cost, and outcomes dynamically

Insight:
Filter Actions help users explore specific healthcare data interactively and quickly.

Business Value:
Helps hospital management identify department performance, patient trends, and critical cases more efficiently.

### Question 3: Create a Bar Chart showing Total Patients by Hospital Region. Add this as a Story Point with proper title.

-->

Answer:

This task creates a Bar Chart to analyze total patients by hospital region and adds it as a Tableau Story Point.

Steps:

1. Drag Hospital_Region → Rows  
2. Drag COUNTD(Patient_ID) → Columns  
3. Select Bar Chart from Show Me  
4. Sort chart in descending order  
5. Drag COUNTD(Patient_ID) → Label  
6. Format labels and chart title properly  
7. Rename worksheet as “Patient Distribution by Region”  
8. Add worksheet to Story  
9. Set Story Point title as “Total Patients by Hospital Region”  
10. Adjust formatting and layout for better readability  

Insight:  
The visualization shows which hospital regions have the highest patient count.

Business Value:  
Helps hospital management improve resource planning and regional healthcare management.

### Question 7: Doctor Performance & Patient KPI Analysis

Task Part A – Create a Ranked Bar Chart showing Top Doctors by Total Treatment Cost handled.

Task Part B – Create a KPI visualization showing Total Patients Treated.

-->

Answer:

This task creates a ranked doctor performance chart and a KPI card to monitor total patients treated.

Steps:

Part A – Doctor Performance

1. Drag Doctor_Name → Rows  
2. Drag SUM(Treatment_Cost) → Columns  
3. Select Bar Chart from Show Me  
4. Sort doctors in descending order by SUM(Treatment_Cost)  
5. Drag SUM(Treatment_Cost) → Label  
6. Apply Top Filter → Top → By Field → Top 10 by SUM(Treatment_Cost)  
7. Format Treatment_Cost as Currency  
8. Customize colors and rename worksheet as “Top Doctors by Treatment Cost”  

Part B – KPI Creation

9. Create a new worksheet  
10. Drag COUNTD(Patient_ID) → Text on Marks card  
11. Change Marks type to Text  
12. Increase font size to highlight KPI value  
13. Add title as “Total Patients Treated”  
14. Format alignment and KPI display for dashboard use  

Insight:  
The ranked chart identifies top doctors by treatment value, while KPI highlights overall patient volume.

Business Value:  
Helps hospital management track doctor contribution and monitor overall patient treatment performance.
