### Question 1: Create a Card Visual to show Total Sales.

-->  

Answer:  

Steps:

1. Open **Power BI Desktop** and go to **Report View**.  
2. In the **Visualizations pane**, select the **Card visual**.  
3. Drag the **Sales** field into the **Values** section.  
4. Power BI will automatically calculate the **Sum of Sales**.  
5. Resize and position the card on the report canvas.  
6. (Optional) Format the card → turn on **Title** and rename it to **Total Sales**.  

This visual displays the total sales value in a single number.

👉 See report: https://drive.google.com/file/d/19v3b2Shqu8g4ASRmJXPEcdBxo4ipknID/view?usp=sharing

### Question 2: Create a Donut Chart showing Total Sales by Ship Mode and which ship mode generates the highest sales?

-->  

Answer:  

Steps:

1. Go to **Report View**.  
2. Select **Donut Chart** from Visualizations.  
3. Drag **Ship Mode** to **Legend**.  
4. Drag **Sales** to **Values**.  
5. Enable **Data Labels** for better visibility.  
6. Identify the segment with the largest value (highest sales).  

The ship mode with the largest slice represents the highest sales.

### Question 3: Create a Line Chart showing the Sales Trend over time (by Month and Year). Is there any noticeable seasonal pattern in sales?

-->  

Answer:  

Steps:

1. Select **Line Chart** from Visualizations.  
2. Drag **Order Date** to **X-axis**.  
3. Set it to **Month-Year hierarchy**.  
4. Drag **Sales** to **Values**.  
5. Format the chart (title, labels if needed).  

You can observe peaks and dips to identify seasonal trends.

### Question 5: Create a Matrix Visual to display Sales by Month and Category and identify the month with the highest sales.

-->  

Answer:  

Steps:

1. Select **Matrix visual**.  
2. Drag **Month (from Order Date)** to **Rows**.  
3. Drag **Category** to **Columns**.  
4. Drag **Sales** to **Values**.  
5. Enable **Conditional Formatting** (optional) to highlight high values.  
6. Identify the row (month) with the highest total sales.

### Question 6: Create a Decomposition Tree for Sales Breakdown.

-->  

Answer:  

Steps:

1. Select **Decomposition Tree** visual.  
2. Drag **Sales** to **Analyze**.  
3. Drag fields like **Category, Region, Ship Mode** to **Explain By**.  
4. Expand branches to analyze how sales are distributed.  
5. Use drill-down to explore contributing factors.  

This helps break down total sales into contributing dimensions.

### Question 7: Create a Scatter Chart to compare Discount vs. Profit by Sub-Category. Are there any outliers where high discounts resulted in negative profit?

-->  

Answer:  

Steps:

2. Drag **Discount** to **X-axis**.  
3. Drag **Profit** to **Y-axis**.  
4. Drag **Sub-Category** to **Legend**.  
5. (Optional) Add **Sales** to **Size**.  
6. Analyze points where **high discount + negative profit** appear.  

Outliers can be identified where points fall in high discount and low/negative profit area.

### Question 10: Create a Dual-Axis Line Chart to show Total Sales and Profit Over Time (by Month).

-->  

Answer:  

Steps:

1. Select **Line and Clustered Column Chart** (for dual axis).  
2. Drag **Order Date (Month)** to **X-axis**.  
3. Drag **Sales** to **Column Values**.  
4. Drag **Profit** to **Line Values**.  
5. Enable **Secondary Axis** for Profit.  
6. Format chart (titles, labels).  

This helps compare sales and profit trends over time.
