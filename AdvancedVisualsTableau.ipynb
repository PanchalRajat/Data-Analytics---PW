### Question 1: Create a Heatmap to analyze Treatment Cost by Department and Region.

-->

Answer:

This task is about analyzing how treatment costs vary across different departments and regions using a Heatmap.

Steps:

1. Drag "Department" to Rows  
2. Drag "Region" to Columns  
3. Drag "TreatmentCost" to Color in the Marks card  
4. Ensure aggregation is set to SUM (Sum of TreatmentCost)  
5. Change Marks type to "Square"   
6. Drag "TreatmentCost" to Label to display values inside cells  
7. Click on Color → Edit Colors → choose a gradient color palette  
8. Ensure darker colors represent higher treatment cost   
9. Analyze:
    - High-cost departments by region
    - Regions with overall high treatment expenses

### Question 4: Create a Butterfly Chart comparing Male vs Female patient count by Department.

-->

Answer:

This task is about creating a Butterfly Chart to compare male and female patient counts across departments.

Steps:

1. Create a Calculated Field named "Male Count"

2. Use logic to assign negative values for male patients

3. Create another Calculated Field named "Female Count"

4. Use logic to assign positive values for female patients

5. Drag "Department" to Rows

6. Drag "Male Count" to Columns

7. Drag "Female Count" to Columns beside Male Count

8. Ensure aggregation is set to SUM for both fields

9. Change Marks type to Bar

10. Use different colors:
    - Male → one color
    - Female → another color

11. Format axis:
    - Male bars should extend left
    - Female bars should extend right

12. Show labels for patient counts

13. Analyze:
    - Departments with more male patients
    - Departments with more female patients

Code:

Male Count:

```
IF [Gender] = "Male" THEN -1
ELSE 0
END
```

Female Count:
```
IF [Gender] = "Female" THEN 1
ELSE 0
END
```

-->

12. Analyze:
