### Question 1: Explain why publishing and sharing reports is considered more important than just building dashboards in Power BI Desktop.

-->  

Answer:  

Building dashboards in Power BI Desktop is only the first step. The real value comes when reports are **shared with users for decision-making**.

Publishing and sharing are more important because:
- Reports become accessible to **teams and stakeholders**
- Enables **real-time access and collaboration**
- Helps in **better and faster decision-making**
- Makes dashboards useful in real business scenarios

Steps:

1. Create report in **Power BI Desktop**.  
2. Click on **Publish**.  
3. Select a **workspace**.  
4. Share the report with users in Power BI Service.

### Question 4: List and explain the steps involved in publishing a report from Power BI Desktop to Power BI Service in the correct sequence.

-->  

Answer:  

Publishing a report means transferring it from Power BI Desktop to the cloud (Power BI Service) so it can be accessed and shared.

Steps (Correct Sequence):

1. Open the report in **Power BI Desktop**.  
2. Click on the **Home tab → Publish** button.  
3. Sign in to your **Power BI account** (if not already signed in).  
4. Select the **destination workspace**.  
5. Click **Select/Publish**.  
6. Wait for the confirmation message (**Publish successful**).  
7. Open **Power BI Service (browser)** to verify the report.  

This process makes the report available online for sharing and collaboration.

### Question 5: Differentiate between Report and Dashboard in Power BI based on structure and usage.

-->  

Answer:  

A **Report** and a **Dashboard** are both used for data visualization but differ in structure and purpose.

- **Report**:
  - Can have **multiple pages**
  - Created in **Power BI Desktop**
  - Contains detailed visuals and analysis
  - Used for **deep data exploration**

- **Dashboard**:
  - Single page (canvas)
  - Created in **Power BI Service**
  - Contains pinned visuals from reports
  - Used for **quick overview of key metrics**

Steps (Usage):

1. Create report in Desktop.  
2. Publish to Service.  
3. Pin visuals to create dashboard.  

Reports = detailed analysis  
Dashboards = summary view

### Question 6: Explain the difference between Direct Report Sharing, Workspace Access, and Power BI Apps. Also mention one use case for each.

-->  

Answer:  

Power BI provides different ways to share reports depending on user needs.

1. **Direct Report Sharing**  
   - Share a specific report directly with users  
   - Users get access to only that report  

   Use case: Sharing a sales report with a manager  

2. **Workspace Access**  
   - Users are given access to an entire workspace  
   - Can view, edit, or manage content based on role  

   Use case: Team collaboration on reports  

3. **Power BI Apps**  
   - Packaged collection of reports and dashboards  
   - Shared with a large audience in a structured way  

   Use case: Company-wide dashboard distribution  

Steps:

1. Publish report.  
2. Choose sharing method (Direct / Workspace / App).  
3. Assign permissions.  

Each method differs in control and scale of sharing.

### Question 7: Describe the four workspace roles available in Power BI and explain how access control is managed using these roles.

-->  

Answer:  

Power BI uses workspace roles to control access and permissions for users.

The four roles are:

1. **Admin**  
   - Full control over workspace  
   - Can add/remove users, publish, edit, delete  

2. **Member**  
   - Can create, edit, and publish content  
   - Cannot manage permissions fully  

3. **Contributor**  
   - Can create and edit reports  
   - Cannot share or manage access  

4. **Viewer**  
   - Can only view reports and dashboards  
   - No editing rights  

Access Control:

- Permissions are assigned based on roles  
- Each role defines what actions a user can perform  
- Helps ensure **data security and controlled access**

Steps:

1. Open workspace in Power BI Service.  
2. Click **Access / Manage Access**.  
3. Add users.  
4. Assign roles (Admin, Member, Contributor, Viewer).  

This ensures proper control over who can view or modify reports.

### Question 8: Why are Power BI Apps considered more secure and scalable than direct sharing? Explain this from a governance and enterprise usage perspective.

-->  

Answer:  

Power BI Apps are designed for **enterprise-level distribution**, making them more secure and scalable compared to direct sharing.

- **Security (Governance):**
  - Apps provide **controlled access** to a fixed set of reports
  - Users cannot accidentally edit or misuse data
  - Centralized management ensures consistency

- **Scalability:**
  - Apps can be shared with **large groups or entire organizations**
  - No need to share reports individually
  - Updates are automatically reflected for all users

Steps:

1. Publish reports to a workspace  
2. Create an **App** from the workspace  
3. Define audience and permissions  
4. Distribute the app  

Thus, Apps provide **better control, security, and large-scale sharing**.

### Question 9: Explain Row-Level Security (RLS) in Power BI. How does RLS ensure that different users see different data using the same report?

-->  

Answer:  

Row-Level Security (RLS) is a feature that **restricts data access at the row level**, ensuring users see only relevant data.

How it works:
- Rules (filters) are defined on data tables
- These rules are applied based on user roles
- Same report shows **different data for different users**

Example:
- Manager sees all data  
- Salesperson sees only their region  

Steps:

1. Go to **Model View → Manage Roles**  
2. Define filters (e.g., Region = "North")  
3. Assign users to roles in Power BI Service  
4. Publish report  

RLS ensures **data privacy and controlled visibility** within the same report.

### Question 10: Assume an organization shares reports without using Apps, workspace roles, or RLS. Discuss any three risks or issues and explain how Power BI best practices help prevent them.

-->  

Answer:  

Without proper sharing mechanisms, several risks can arise:

1. **Data Security Risk**  
   - Sensitive data may be visible to unauthorized users  
   - No control over who sees what  

   Prevention:
   - Use **RLS** to restrict data access  

2. **Lack of Governance**  
   - No structured control over reports  
   - Multiple versions of the same report may exist  

   Prevention:
   - Use **Workspaces and Apps** for centralized management  

3. **Data Misuse or Misinterpretation**  
   - Users may access incomplete or irrelevant data  
   - Can lead to wrong decisions  

   Prevention:
   - Use **controlled sharing and role-based access**  

Steps:

1. Use Workspaces for structured collaboration  
2. Create Apps for distribution  
3. Apply RLS for data security  

These best practices ensure **secure, organized, and reliable reporting**.
