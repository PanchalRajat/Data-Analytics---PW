### Q1. Creating a DataFrame

-->

Answer:
A DataFrame is a two-dimensional data structure provided by the Pandas library that stores data in rows and columns. It is one of the most important data structures used in data analysis and data science because it allows data to be organized, manipulated, filtered, and analyzed efficiently.

In this question, a DataFrame is created using a Python dictionary. Each key in the dictionary represents a column name, while the corresponding values represent the data stored in that column. The DataFrame contains four columns: Product, Region, Sales, and Quantity. Five sample records are added to demonstrate how data can be stored in a tabular format.

Creating a DataFrame from a dictionary is one of the simplest and most commonly used methods when data is already available in Python. Once the DataFrame is created, it can be displayed to verify that the data has been loaded correctly.

```python
import pandas as pd

data = {
    'Product': ['Laptop', 'Mouse', 'Keyboard', 'Monitor', 'Printer'],
    'Region': ['North', 'South', 'East', 'West', 'North'],
    'Sales': [55000, 1200, 2500, 15000, 8000],
    'Quantity': [5, 20, 15, 8, 4]
}

df = pd.DataFrame(data)

print(df)
```

### Q2. Data Inspection

-->

Answer:
Data inspection is the process of understanding the structure and contents of a dataset before performing any analysis. It helps identify the number of rows and columns, data types, missing values, and sample records. Proper inspection ensures that the dataset is correctly loaded and ready for further processing.

The `head()` function is used to display the first five rows of the dataset, which provides an overview of how the data is organized. The `tail()` function displays the last five rows, allowing us to inspect records from the end of the dataset. The `info()` function provides detailed information about the dataset, including column names, data types, and non-null values. Finally, the `shape` attribute returns the total number of rows and columns present in the dataset.

Performing these inspection tasks is considered a best practice because it helps analysts understand the dataset before applying transformations, filtering, or calculations.

```python
import pandas as pd

df = pd.read_csv('Retail_dataset.csv')

print("First 5 Rows:")
print(df.head())

print("\nLast 5 Rows:")
print(df.tail())

print("\nDataset Information:")
print(df.info())

print("\nDataset Shape:")
print(df.shape)
```

### Q3. Column Selection

-->

Answer:
Column selection is used when only specific columns from a dataset are required for analysis. Instead of working with the entire dataset, selecting relevant columns helps reduce complexity and improves readability.

In this task, only the Product, Region, and Sales columns are selected from the dataset. ### Q4. Indexing using .loc

-->

Answer:
The `.loc[]` method in Pandas is used for label-based indexing. It allows users to access specific rows and columns using their index labels and column names. This method is particularly useful when the exact row index and column name are known.

For this question, the entire row with index 10 is displayed. This allows us to view all information associated with that particular record. Next, only the Sales value from the row with index 20 is retrieved. This demonstrates how `.loc[]` can be used to access a single value from a specific row and column.

Label-based indexing provides precise control over data retrieval and is widely used when working with structured datasets.

```python
import pandas as pd

df = pd.read_csv('Retail_dataset.csv')

print("Row with Index 10:")
print(df.loc[10])

print("\nSales Value at Index 20:")
print(df.loc[20, 'Sales'])
```

### Q5. Indexing using .iloc

-->

Answer:
The `.iloc[]` method is used for integer position-based indexing in Pandas. Unlike `.loc[]`, which relies on labels, `.iloc[]` accesses data based on the numerical position of rows and columns.

In this task, the first six rows and first four columns are displayed using `.iloc[]`. This helps examine a specific section of the dataset based on row and column positions. After that, the Sales value from the third row is extracted using its position in the dataset.

Position-based indexing is particularly useful when row labels are not important or when analysts want to access data using numerical coordinates.

```python
import pandas as pd

df = pd.read_csv('Retail_dataset.csv')

print("First 6 Rows and First 4 Columns:")
print(df.iloc[:6, :4])

print("\nSales Value from 3rd Row:")
print(df.iloc[2]['Sales'])
```

### Q6. Row Slicing

-->

Answer:
Row slicing is used to retrieve a specific range of rows from a dataset. It helps analysts focus on a particular section of data without displaying the entire dataset. This technique is especially useful when working with large datasets because it allows quick inspection of selected records.

In this task, the first 10 rows of the dataset are displayed to get an overview of the initial records. After that, rows 15 to 25 are displayed to examine data from the middle section of the dataset. This helps in understanding how records are distributed throughout the dataset and allows users to inspect specific portions of data more efficiently.

```python
import pandas as pd

df = pd.read_csv('Retail_dataset.csv')

print("First 10 Rows:")
print(df.head(10))

print("\nRows 15 to 25:")
print(df.iloc[15:26])
```

### Q7. Filtering Data

-->

Answer:
Filtering is the process of selecting only those rows that satisfy a specific condition. It allows users to focus on relevant records and perform targeted analysis.

In this question, two filters are applied. The first filter displays all rows where the Region is "East". This helps analyze sales data specifically for the East region. The second filter displays all rows where the Sales value is greater than 1500. This helps identify transactions that generated higher revenue and can be useful for studying top-performing sales records.

Filtering is one of the most frequently used operations in data analysis because it enables users to extract meaningful information from large datasets.

```python
import pandas as pd

df = pd.read_csv('Retail_dataset.csv')

print("Rows where Region = East:")
print(df[df['Region'] == 'East'])

print("\nRows where Sales > 1500:")
print(df[df['Sales'] > 1500])
```

### Q8. Multiple Condition Filtering

-->

Answer:
Sometimes a single condition is not enough to identify the required records. In such cases, multiple conditions can be combined using logical operators such as AND (`&`) and OR (`|`).

For this task, two conditions are applied simultaneously. The first condition selects records where the Region is "West", while the second condition selects records where the Sales value is greater than 1200. Using the AND operator ensures that only rows satisfying both conditions are displayed.

This type of filtering is commonly used in business analysis to identify specific groups of records that meet multiple criteria at the same time.

```python
import pandas as pd

df = pd.read_csv('Retail_dataset.csv')

filtered_data = df[
    (df['Region'] == 'West') &
    (df['Sales'] > 1200)
]

print(filtered_data)
```

### Q9. Descriptive Statistics

-->

Answer:
Descriptive statistics provide a summary of the numerical characteristics of a dataset. Instead of examining individual records, descriptive statistics help understand the overall distribution and behavior of the data.

The `describe()` function generates important statistical measures such as count, mean, standard deviation, minimum value, maximum value, and quartiles. These statistics help analysts understand the central tendency, spread, and range of the data.

From the generated summary statistics, we can identify the mean sales value, maximum quantity sold, and minimum sales value. These measures provide useful insights into sales performance and customer purchasing behavior.

```python
import pandas as pd

df = pd.read_csv('Retail_dataset.csv')

stats = df.describe()

print("Summary Statistics:")
print(stats)

print("\nMean Sales Value:")
print(stats.loc['mean', 'Sales'])

print("\nMaximum Quantity Sold:")
print(stats.loc['max', 'Quantity'])

print("\nMinimum Sales Value:")
print(stats.loc['min', 'Sales'])
```

### Q10. Aggregation Analysis

-->

Answer:
Aggregation is the process of combining multiple values to produce a summary result. It is widely used in data analysis to understand overall performance and identify important trends within a dataset.

In this question, aggregation functions are applied to the Sales column. The total sales value is calculated using the sum of all sales records. The average sales value is calculated using the mean function, which represents the typical sales amount. The maximum sales value identifies the highest sale recorded in the dataset, while the minimum sales value identifies the lowest sale.

These aggregated measures provide a quick overview of the overall sales performance and help businesses evaluate revenue trends and make informed decisions.

```python
import pandas as pd

df = pd.read_csv('Retail_dataset.csv')

total_sales = df['Sales'].sum()
average_sales = df['Sales'].mean()
maximum_sales = df['Sales'].max()
minimum_sales = df['Sales'].min()

print("Total Sales:", total_sales)
print("Average Sales:", average_sales)
print("Maximum Sales:", maximum_sales)
print("Minimum Sales:", minimum_sales)
```
