### Q3. Customer Purchase Analysis

Identify the top 5 customers based on total sales.

Hint: Use GroupBy with aggregation.

-->

Answer:
Customer purchase analysis is performed to identify the customers who generate the highest revenue for the business. Understanding customer spending behavior is important because a small number of customers often contribute a large percentage of total sales.

Using the `groupby()` function, all transactions belonging to the same customer are combined together. The `sum()` function is then applied to calculate the total sales generated by each customer across all their purchases. Finally, the results are sorted in descending order to identify the highest-spending customers.

The top five customers represent the most valuable customers in the dataset. Businesses can use this information to improve customer retention, create loyalty programs, and provide personalized services to high-value customers.

```python
import pandas as pd

df = pd.read_csv('Advanced_pandas_sales_dataset.csv')

top_customers = (
    df.groupby('Customer')['Sales']
    .sum()
    .sort_values(ascending=False)
    .head(5)
)

print("Top 5 Customers Based on Total Sales:")
print(top_customers)
```

### Q4. Regional Sales Performance

Calculate the total sales generated by each region.

Then identify:

- Which region generated the highest revenue?

-->

Answer:
Regional sales analysis helps organizations evaluate business performance across different geographical locations. By calculating total sales for each region, businesses can determine which markets contribute the most revenue and which areas may require additional attention.

The dataset is grouped by Region, and the Sales column is aggregated using the `sum()` function. This calculates the total revenue generated by each region. After calculating regional sales totals, the region with the highest revenue is identified using the `idxmax()` function.

This analysis helps management understand regional performance, allocate resources effectively, and make informed business decisions regarding expansion and marketing investments.

```python
import pandas as pd

df = pd.read_csv('Advanced_pandas_sales_dataset.csv')

regional_sales = df.groupby('Region')['Sales'].sum()

print("Total Sales by Region:")
print(regional_sales)

highest_region = regional_sales.idxmax()
highest_revenue = regional_sales.max()

print("\nRegion with Highest Revenue:", highest_region)
print("Highest Revenue:", highest_revenue)
```

### Q5. Product Performance Analysis

Calculate the total quantity sold for each product.

Which product appears to be the most popular?

-->

Answer:
Product performance analysis is used to determine which products are most frequently purchased by customers. One effective measure of product popularity is the total quantity sold because it reflects actual customer demand.

The dataset is grouped by Product, and the Quantity column is summed for each product. This provides the total number of units sold for every product in the dataset. The product with the highest total quantity sold is considered the most popular because it has been purchased more often than any other product.

Understanding product popularity helps businesses manage inventory levels, forecast future demand, and prioritize marketing efforts for high-performing products.

```python
import pandas as pd

df = pd.read_csv('Advanced_pandas_sales_dataset.csv')

product_quantity = (
    df.groupby('Product')['Quantity']
    .sum()
    .sort_values(ascending=False)
)

print("Total Quantity Sold for Each Product:")
print(product_quantity)

print("\nMost Popular Product:")
print(product_quantity.idxmax())
```

### Q7. Pivot Table Creation

Create a pivot table showing:

- Rows → Region
- Columns → Product
- Values → Total Sales

Which region sells the most laptops?

-->

Answer:
A Pivot Table is a powerful data summarization tool in Pandas that allows data to be grouped and analyzed across multiple dimensions. It helps transform large datasets into a structured format that is easier to interpret and analyze.

In this question, a pivot table is created where Region is used as the row index, Product is used as the column header, and the Sales column is aggregated using the sum function. This provides a clear view of how much revenue each product generates in each region.

To identify which region sells the most laptops, the Laptop column from the pivot table can be examined and the region with the highest sales value can be identified using the `idxmax()` function.

This analysis helps businesses understand regional demand for specific products and identify strong-performing markets.

```python
import pandas as pd

df = pd.read_csv('Advanced_pandas_sales_dataset.csv')

pivot_table = pd.pivot_table(
    df,
    values='Sales',
    index='Region',
    columns='Product',
    aggfunc='sum'
)

print("Pivot Table:")
print(pivot_table)

print("\nRegion Selling the Most Laptops:")
print(pivot_table['Laptop'].idxmax())
```

### Q8. Profitability Analysis

Calculate the average profit for each product category.

Which category is most profitable?

-->

Answer:
Profitability analysis helps businesses determine which product categories generate the highest returns. While sales indicate revenue, profit provides a clearer picture of actual business performance because it accounts for costs and discounts.

The dataset is grouped by Category, and the average profit is calculated using the `mean()` function. This shows the typical profit generated by products within each category. The category with the highest average profit is considered the most profitable because it consistently generates better returns compared to other categories.

Understanding category profitability helps businesses focus resources on high-performing product groups and improve decision-making related to inventory, pricing, and marketing.

```python
import pandas as pd

df = pd.read_csv('Advanced_pandas_sales_dataset.csv')

avg_profit = (
    df.groupby('Category')['Profit']
    .mean()
    .sort_values(ascending=False)
)

print("Average Profit by Category:")
print(avg_profit)

print("\nMost Profitable Category:")
print(avg_profit.idxmax())
```

### Q9. Discount Impact Study

Analyze the relationship between discount and profit.

Tasks:

- Calculate the average profit for each discount level
- Identify whether higher discounts lead to lower profit

-->

Answer:
Discounts are often used to increase sales volume and attract customers. However, offering larger discounts may reduce profit margins. Therefore, it is important to study the relationship between discount levels and profitability.

In this analysis, the dataset is grouped by Discount level and the average profit is calculated for each group. This allows us to observe how profit changes as discounts increase.

If average profit decreases as discount levels become higher, it suggests that larger discounts are reducing profitability. On the other hand, if profit remains stable or increases, it may indicate that increased sales volume is compensating for the discount offered.

This type of analysis helps businesses determine the optimal discount strategy that balances customer attraction and profitability.

```python
import pandas as pd

df = pd.read_csv('Advanced_pandas_sales_dataset.csv')

discount_profit = (
    df.groupby('Discount')['Profit']
    .mean()
    .sort_index()
)

print("Average Profit by Discount Level:")
print(discount_profit)
```

### Q10. End-to-End Data Wrangling Task

Perform the following workflow:

- Filter the dataset where Sales > 3000
- Group data by Region and Product
- Calculate total sales
- Reshape the result into a pivot table

Explain the business insights obtained from this analysis.

-->

Answer:
Data wrangling is the process of transforming raw data into a structured format that is easier to analyze and interpret. In this task, multiple Pandas operations are combined to perform a complete business analysis workflow.

First, the dataset is filtered to include only records where Sales are greater than 3000. This ensures that the analysis focuses only on high-value transactions. Next, the filtered data is grouped by Region and Product, and total sales are calculated for each combination. Finally, the grouped result is reshaped into a pivot table to create a clear summary of product performance across regions.

Business Insights:
- High-value products become easier to identify within each region.
- Strong-performing regions can be identified by comparing total sales.
- Product demand patterns across regions become more visible.
- Management can focus marketing efforts on products generating high revenue.
- Inventory planning can be optimized by understanding which products perform best in each region.
- Underperforming products or regions can be identified and targeted for improvement.

This workflow demonstrates how filtering, grouping, aggregation, and reshaping can be combined to generate meaningful business insights from raw sales data.

```python
import pandas as pd

df = pd.read_csv('Advanced_pandas_sales_dataset.csv')

filtered_df = df[df['Sales'] > 3000]

grouped_sales = (
    filtered_df.groupby(['Region', 'Product'])['Sales']
    .sum()
    .reset_index()
)

pivot_result = grouped_sales.pivot(
    index='Region',
    columns='Product',
    values='Sales'
)

print("Filtered Data:")
print(filtered_df.head())

print("\nGrouped Sales:")
print(grouped_sales)

print("\nPivot Table:")
print(pivot_result)
```
