### Question 2: Explain the difference between List and Tuple in Python.

-->

Answer:

List and Tuple are Python data structures used to store multiple values in a single variable.

Steps:

1. A List is created using square brackets [ ].
2. A Tuple is created using parentheses ( ).
3. Lists are mutable, meaning elements can be added, removed, or modified.
4. Tuples are immutable, meaning values cannot be changed after creation.
5. Lists are generally used for dynamic data, while Tuples are used for fixed data.

Insight:
The main difference is that Lists can be modified, whereas Tuples cannot.

Business Value:
Selecting the correct data structure improves program performance and data management.

### Question 5: Write a Python program to check whether a number is even or odd using conditional statements.

-->

Answer:

This program determines whether a given number is even or odd using an if-else condition.

Steps:

1. Accept a number from the user using input().
2. Convert the input value into an integer.
3. Use the modulus operator (%) to divide the number by 2.
4. Check whether the remainder is equal to 0.
5. If the remainder is 0, display "Even Number".
6. Otherwise, display "Odd Number".

```python
num = int(input("Enter a number: "))

if num % 2 == 0:
    print("Even Number")
else:
    print("Odd Number")
```

Insight:
An even number is completely divisible by 2, while an odd number leaves a remainder of 1.

Business Value:
Conditional logic is widely used in validation systems and business rule implementation.

### Question 6: Write a program to print numbers from 1 to 10 using a loop.

-->

Answer:

This program uses a loop to print numbers from 1 to 10 automatically.

Steps:

1. Use a for loop to repeat a task multiple times.
2. Use the range() function to generate numbers.
3. Set the starting value as 1.
4. Set the ending value as 11 because the last value is excluded.

### Question 8: Write a Python program to remove duplicate values from a list using a set.

-->

Steps:

2. Convert the list into a set using set().
3. Store the unique values in a new variable.

### Question 10: Write a Python program to count how many times a number appears in a list.

Example List: [2,3,4,2,5,2]

-->

Answer:

This program counts the frequency of a specific number in a list.

Steps:

1. Create the given list.
2. Choose the number to search for.
3. Use the count() method on the list.
4. Store the result in a variable.
5. Print the total occurrences.

```python
numbers = [2, 3, 4, 2, 5, 2]

count = numbers.count(2)

print("Occurrences:", count)
```

Insight:
The count() method returns the total number of times a value appears in a list.

Business Value:
Useful for frequency analysis, inventory tracking, and data reporting.
