### Question 3: Replace missing values in the Quantity and Price columns with their respective mean values using Pandas.

-->

### Question 7: Group the dataset by Product Category and calculate the total revenue generated by each category.

-->

Answer:

This task calculates category-wise revenue.

Steps:

1. Group data by Product Category.
2. Sum Revenue values.
3. Display results.

```python
category_revenue = df.groupby("Product Category")["Revenue"].sum()

print(category_revenue)
```

Insight:
Shows revenue contribution of each category.

Business Value:

### Question 8: Based on the revenue calculated for each product category, identify the top 3 categories with the highest revenue and the bottom 3 categories with the lowest revenue.

-->

Answer:

This task identifies best and worst performing categories.

Steps:

1. Sort category revenue in descending order.
2. Select top 3 categories.

### Question 9: Group the dataset by Region and calculate the total revenue generated in each region. Identify the region with the highest revenue and the region with the lowest revenue.

-->

Answer:

This task analyzes revenue by region.

Steps:

1. Group dataset by Region.
2. Sum Revenue values.
3. Find highest revenue region.
4. Find lowest revenue region.

```python
region_revenue = df.groupby("Region")["Revenue"].sum()

print(region_revenue)

print("Highest Revenue Region:")
print(region_revenue.idxmax())

print("Lowest Revenue Region:")
print(region_revenue.idxmin())
```

Insight:
Shows regional sales performance.

Business Value:

### Question 10: Extract the month from the Date column using Pandas and group the dataset by month to calculate the total monthly revenue. Also, calculate the mean, median, and standard deviation of the Revenue column using NumPy.

-->

Answer:

This task performs monthly revenue and statistical analysis.

Steps:

1. Convert Date column to datetime.
2. Extract month from Date.
3. Group by Month and calculate revenue.
4. Calculate mean, median, and standard deviation of Revenue.

```python
df["Date"] = pd.to_datetime(df["Date"])

df["Month"] = df["Date"].dt.month

monthly_revenue = df.groupby("Month")["Revenue"].sum()

print(monthly_revenue)

print("Mean:", np.mean(df["Revenue"]))
print("Median:", np.median(df["Revenue"]))
print("Standard Deviation:", np.std(df["Revenue"]))
```

Insight:
Shows monthly sales trends and revenue distribution.

Business Value:
Supports forecasting and business planning.
