### Question 1: What is Power BI and why is it used in businesses?

-->  

Answer:  
Power BI is a business intelligence (BI) tool developed by Microsoft that is used to analyze data and create interactive visualizations such as dashboards and reports. It allows users to connect to different data sources, transform data, and present insights in a visual format.

Power BI is widely used in businesses because it helps convert raw data into meaningful information. It enables organizations to track performance, monitor key metrics, and make data-driven decisions.

Some key reasons why businesses use Power BI include:
- **Data visualization** – Converts complex data into easy-to-understand charts and graphs  
- **Real-time insights** – Helps monitor business performance with up-to-date data  
- **Integration** – Connects with multiple data sources like Excel, SQL databases, and cloud services  
- **Decision making** – Supports better and faster business decisions through data analysis  

Overall, Power BI helps businesses understand their data effectively and improve operational efficiency.

### Question 2: Name and explain the three main components of Power BI.

-->  

Answer:  
The three main components of Power BI are:

1. **Power BI Desktop** – This is a Windows-based application used to create reports and dashboards. Users can import data from different sources, clean and transform it, and design visualizations.

2. **Power BI Service** – This is a cloud-based platform where reports and dashboards created in Power BI Desktop are published and shared. It allows users to access reports online, collaborate with others, and view data from anywhere.

3. **Power BI Mobile** – This is a mobile application available for smartphones and tablets. It allows users to view and interact with reports and dashboards on the go.

These components work together to help users create, share, and access business insights easily.

### Question 3: Explain the Power BI workflow.

-->  

Answer:  
The Power BI workflow describes the steps involved in turning raw data into meaningful insights through reports and dashboards.

First, data is **collected from different sources** such as Excel files, databases, or cloud services using Power BI Desktop.

Next, the data is **cleaned and transformed** using Power Query. This step involves removing errors, handling missing values, and formatting data properly.

After that, the cleaned data is **modeled** by creating relationships between tables and defining calculations using DAX (Data Analysis Expressions).

Then, users create **visualizations and reports** by adding charts, tables, and other visuals to represent the data clearly.

Finally, the report is **published to Power BI Service**, where it can be shared with others, accessed online, and viewed on mobile devices.

This workflow helps convert raw data into interactive dashboards for better decision-making.

### Question 6: Define Data View, Report View, and Model View. Explain the purpose of each view.

-->  

Answer:  
Power BI Desktop provides three main views that help users work with data effectively:

1. **Data View** – This view displays the dataset in a tabular format (like a table). It is used to inspect data, create calculated columns, and understand the structure of the dataset.

2. **Report View** – This is the main working area where users create visualizations such as charts, tables, and dashboards. It is used to design reports and present insights from the data.

3. **Model View** – This view shows the relationships between different tables in the form of a diagram. It is used to create and manage relationships, define data models, and ensure proper data connections.

Each view serves a specific purpose: Data View for examining data, Report View for visualization, and Model View for managing relationships and data structure.

### Question 7: Discuss the different data sources that Power BI supports.

-->  

Answer:  
Power BI supports a wide range of data sources, allowing users to connect and analyze data from different platforms.

One common type is **file-based sources**, such as Excel, CSV, and text files. These are widely used for importing structured data.

Another type is **database sources**, including SQL Server, MySQL, Oracle, and PostgreSQL. Power BI can directly connect to these databases and retrieve data using queries.

Power BI also supports **cloud-based sources** such as Azure, Google Analytics, SharePoint, and online services. These sources allow access to data stored in cloud platforms.

Additionally, it can connect to **web and API sources**, where data can be fetched from websites or external applications through APIs.

Finally, Power BI supports **big data and data warehouse systems** such as Azure Synapse, Snowflake, and other large-scale data platforms.

This flexibility allows Power BI to integrate data from multiple sources and create comprehensive reports and dashboards.

### Question 8: Split Owner Name to create two new columns as First Name and Last Name

-->  

Answer:  
To split the Owner Name column into First Name and Last Name in Power BI, the following steps can be performed using Power Query:

1. Open **Power Query Editor** by clicking on **Transform Data**.

2. Select the **Owner_Name** column.

3. Go to the **Home** or **Transform** tab and click on **Split Column**.

4. Choose **By Delimiter**.

5. Select **Space** as the delimiter (since first and last names are separated by a space).

6. Click **OK**.

7. Power BI will create two new columns (e.g., Owner_Name.1 and Owner_Name.2).

8. Rename the columns as **First_Name** and **Last_Name**.

This process separates the full name into two individual columns for better analysis and reporting.
