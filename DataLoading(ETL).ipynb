b) Justify your choice  

-->  

Answer:  

**a) Should a Full Load or Incremental Load be used?**  
Incremental Load  

**b) Justify your choice:**  

1. **Efficiency** – Daily sales data is continuously generated, so loading only new or updated records reduces processing time and system load.

2. **Performance** – Incremental load avoids reprocessing the entire dataset every day, making the ETL process faster and more scalable.

3. **Reduced redundancy** – It prevents duplication of already loaded records, especially important since Order_ID is a primary key.

4. **Real-time updates** – Incremental loading allows the system to keep data up-to-date with minimal delay.

Therefore, incremental load is more suitable for daily transactional data.
