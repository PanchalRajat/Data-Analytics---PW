### Question 1: What is the difference between a Basic Filter, Advanced Filter, and a Slicer in Power BI?

-->  

Answer:  

In Power BI, filters are used to control what data is displayed in visuals.

- **Basic Filter** allows users to select specific values (like selecting only 2023).  
- **Advanced Filter** allows applying conditions such as “greater than”, “contains”, etc.  
- **Slicer** is a visual filter placed on the report that allows interactive filtering.

Steps:

1. Open Filters Pane.  
2. Drag a field (like Date or Country).  
3. Choose Basic or Advanced filtering.  
4. Add Slicer visual for interactive filtering.  



### Question 2: Using the Sales dataset, apply a Basic Filter to display only the sales data for the year 2023.

-->  

Answer:  

Basic filtering helps restrict data to specific values like a particular year.

Steps:

1. Select a visual or page.  
2. Drag **Date** into Filters Pane.  
3. Choose **Basic Filtering**.  
4. Select only **2023**.  

Now all visuals will show only 2023 sales.


### Question 3: Create a Pie Chart of Sales by Product and add Slicer that allows users to filter the dataset by Region.

-->

This allows users to see product contribution and filter by region.

Now selecting a region updates the pie chart.


### Question 4: What is a Visual-Level Filter, and how does it differ from Page-Level and Report-Level Filters?

-->  

Answer:  

- **Visual-Level Filter** → applies only to one visual  
- **Page-Level Filter** → applies to all visuals on a page  
- **Report-Level Filter** → applies to all pages  

Steps:

1. Select visual → apply Visual filter.  
2. Use Page filter section.  
3. Use Report filter section.

### Question 6: What is the role of Drill-Down in Power BI?

-->  

Answer:  

Drill-down helps users explore data from summary to detailed levels within the same visual.

Steps:

1. Create hierarchy (Year → Month).  
2. Enable Drill Down icon.  
3. Click to move deeper levels.  

Improves analysis by breaking data step-by-step.
