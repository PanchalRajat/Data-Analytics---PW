### Q1. Create a countplot to show the number of passengers by gender.

-->

Answer:
The dataset contains 891 passengers — 577 males (64.7%) and 314 females (35.3%). Males outnumber females by nearly 2:1. This gender imbalance is historically consistent with the Titanic's passenger manifest, where more male travelers were onboard. This also becomes a critical variable when analyzing survival rates, as females had significantly higher survival priority due to the "women and children first" evacuation policy followed during the disaster.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('Titanic.csv')

plt.figure(figsize=(7, 5))
sns.countplot(data=df, x='Sex', palette=['steelblue', 'salmon'])
plt.title('Number of Passengers by Gender')
plt.xlabel('Gender')
plt.ylabel('Number of Passengers')
plt.xticks([0, 1], ['Male', 'Female'])
for p in plt.gca().patches:
    plt.gca().annotate(int(p.get_height()), (p.get_x() + p.get_width() / 2, p.get_height()),
                       ha='center', va='bottom', fontsize=12)
plt.tight_layout()
plt.show()
```

### Q2. Create a bar chart to show the number of passengers in each passenger class using different colors.

-->

Answer:
Passenger class (Pclass) distribution: Class 1 had 216 passengers (24.2%), Class 2 had 184 passengers (20.6%), and Class 3 had 491 passengers (55.1%). More than half of all passengers traveled in 3rd class, which was the cheapest and most crowded section of the ship. This class distribution directly impacted survival rates — 1st class passengers had better access to lifeboats located on upper decks, while 3rd class passengers faced structural barriers that made evacuation much harder.

```python
pclass_count = df['Pclass'].value_counts().sort_index()

plt.figure(figsize=(7, 5))
colors = ['gold', 'cornflowerblue', 'tomato']
plt.bar(pclass_count.index, pclass_count.values, color=colors, edgecolor='black', width=0.5)
plt.title('Number of Passengers by Passenger Class')
plt.xlabel('Passenger Class')
plt.ylabel('Number of Passengers')
plt.xticks([1, 2, 3], ['Class 1\n(1st)', 'Class 2\n(2nd)', 'Class 3\n(3rd)'])
for i, val in zip(pclass_count.index, pclass_count.values):
    plt.text(i, val + 5, str(val), ha='center', fontsize=12)
plt.tight_layout()
plt.show()
```

### Q3. Create a histogram to show the distribution of passenger age.

-->

Answer:
Age data is available for 714 out of 891 passengers (177 values are missing). The age distribution has a mean of 29.7 years, a median of 28 years, and a skewness of 0.39 — slightly right-skewed but fairly close to normal. Most passengers were between 20 and 40 years old, with the largest group in their mid-to-late 20s. There is a small but visible spike of very young children (under 5), and the distribution thins out significantly after age 50. The oldest passenger was 80 years old. The relatively young average age reflects the demographics of early 20th century transatlantic travelers.

```python
plt.figure(figsize=(9, 5))
plt.hist(df['Age'].dropna(), bins=30, color='steelblue', edgecolor='black', alpha=0.8)
plt.title('Distribution of Passenger Age')
plt.xlabel('Age (years)')
plt.ylabel('Number of Passengers')
plt.axvline(df['Age'].mean(), color='red', linestyle='--', label=f'Mean: {df["Age"].mean():.1f}')
plt.axvline(df['Age'].median(), color='green', linestyle='--', label=f'Median: {df["Age"].median():.1f}')
plt.legend()
plt.tight_layout()
plt.show()
```

### Q4. Create a boxplot to detect outliers in the fare column.

-->

Answer:
The Fare column has a mean of £32.20 and a median of £14.45 — the large gap indicates strong right-skew caused by outliers. Using the IQR method: Q1 = £7.91, Q3 = £31.00, IQR = £23.09, giving an upper bound of £65.63. Any fare above £65.63 is flagged as an outlier — and there are 116 such records (13% of passengers). The maximum fare paid was £512.33, nearly 8x the outlier threshold. These extreme fares belong to wealthy 1st class passengers who booked premium private cabins. The boxplot clearly shows the long right whisker and numerous outlier dots far above the upper fence.

```python
plt.figure(figsize=(8, 5))
sns.boxplot(x=df['Fare'], color='lightcoral')
plt.title('Boxplot of Passenger Fare (Outlier Detection)')
plt.xlabel('Fare (£)')
plt.axvline(65.63, color='blue', linestyle='--', label='Outlier Threshold (£65.63)')
plt.legend()
plt.tight_layout()
plt.show()
```

### Q5a. Create a scatterplot to show the relationship between age and fare.

-->

Answer:
The correlation between Age and Fare is only 0.096 — extremely weak and close to zero. The scatterplot shows a scattered cloud of points with no clear upward or downward trend, confirming that a passenger's age had almost no linear relationship with how much they paid for their ticket. Both young and old passengers can be found paying low fares (3rd class) as well as high fares (1st class). Ticket price was primarily determined by class of travel and cabin type, not the passenger's age. This insight tells us that age should not be used alone as a predictor of fare in any model.

```python
plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, x='Age', y='Fare', alpha=0.5, color='steelblue')
plt.title('Relationship Between Age and Fare')
plt.xlabel('Age (years)')
plt.ylabel('Fare (£)')
plt.tight_layout()
plt.show()

print("Correlation between Age and Fare:", df[['Age','Fare']].corr())
```

### Q5b. Analyze the relationship between family size (SibSp + Parch) and fare.

-->

Answer:
Family size was calculated as SibSp + Parch. Most passengers (537 out of 891, i.e. 60.3%) traveled alone with a family size of 0. The correlation between family size and fare is 0.217 — a weak positive relationship. This means passengers with slightly larger families tended to pay slightly higher total fares, which makes sense as they likely booked multiple tickets or group cabins together. However, passengers with very large families (size 5, 6, or 10) show high variance — some paid very high fares while others paid low fares, suggesting mixed class distribution even among large families. Family size alone is not a strong predictor of fare.

```python
df['Family_Size'] = df['SibSp'] + df['Parch']

plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, x='Family_Size', y='Fare', alpha=0.5, color='darkorange')
plt.title('Relationship Between Family Size and Fare')
plt.xlabel('Family Size (SibSp + Parch)')
plt.ylabel('Fare (£)')
plt.tight_layout()
plt.show()

print("Correlation between Family Size and Fare:", df['Family_Size'].corr(df['Fare']))
```

### Q6. Create a countplot to show survival count by gender.

-->

Answer:
Of 314 female passengers, 233 survived (74.2% survival rate). Of 577 male passengers, only 109 survived (18.9% survival rate). The contrast is stark — females were nearly 4 times more likely to survive than males. This directly reflects the "women and children first" evacuation protocol followed by the Titanic crew. Male passengers, despite being the majority onboard, had the lowest survival priority and faced the least lifeboat access. Gender is therefore one of the strongest predictors of survival in this dataset.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('Titanic.csv')

plt.figure(figsize=(7, 5))
sns.countplot(data=df, x='Sex', hue='Survived', palette={0: 'tomato', 1: 'steelblue'})
plt.title('Survival Count by Gender')
plt.xlabel('Gender')
plt.ylabel('Number of Passengers')
plt.legend(title='Survived', labels=['Did Not Survive', 'Survived'])
plt.tight_layout()
plt.show()
```

### Q7. Create a bar chart to show survival by passenger class.

-->

Answer:
Survival rates by class: Class 1 — 136 out of 216 survived (63.0%), Class 2 — 87 out of 184 survived (47.3%), Class 3 — 119 out of 491 survived (24.2%). There is a clear and strong inverse relationship between passenger class and survival — the higher the class, the better the survival chance. Class 1 passengers had nearly 2.6 times the survival rate of Class 3 passengers. This is because 1st class cabins were located on upper decks with direct lifeboat access, while 3rd class passengers were on lower decks and faced physical barriers and delays during evacuation.

```python
survival_rate = df.groupby('Pclass')['Survived'].mean() * 100

plt.figure(figsize=(7, 5))
colors = ['gold', 'cornflowerblue', 'tomato']
bars = plt.bar(survival_rate.index, survival_rate.values, color=colors, edgecolor='black', width=0.5)
plt.title('Survival Rate by Passenger Class')
plt.xlabel('Passenger Class')
plt.ylabel('Survival Rate (%)')
plt.xticks([1, 2, 3], ['Class 1', 'Class 2', 'Class 3'])
for bar, val in zip(bars, survival_rate.values):
    plt.text(bar.get_x() + bar.get_width()/2, val + 1, f'{val:.1f}%', ha='center', fontsize=12)
plt.tight_layout()
plt.show()
```

### Q8. Create a heatmap to show correlation between numerical variables.

-->

Answer:
Key correlations from the heatmap:

- Pclass vs Fare: -0.549 (strong negative) — higher class number means cheaper fare, i.e. 1st class costs most, 3rd class costs least.
- Pclass vs Survived: -0.338 (moderate negative) — higher class number means lower survival, confirming class-based survival disparity.
- Fare vs Survived: +0.257 (weak positive) — passengers who paid more had slightly better survival chances.
- SibSp vs Parch: +0.415 (moderate positive) — passengers traveling with siblings/spouses also tended to travel with parents/children, i.e. family groups traveled together.
- Age vs Pclass: -0.369 — older passengers tended to be in higher (more expensive) classes.

PassengerId shows near-zero correlation with everything, confirming it is just an ID with no analytical value.

```python
num_cols = ['Survived', 'Pclass', 'Age', 'SibSp', 'Parch', 'Fare']

plt.figure(figsize=(8, 6))
sns.heatmap(df[num_cols].corr(), annot=True, fmt='.2f', cmap='coolwarm',
            linewidths=0.5, square=True)
plt.title('Correlation Heatmap of Numerical Variables')
plt.tight_layout()
plt.show()
```

### Q9. Identify which passenger class paid the highest fare on average.

-->

Answer:
Class 1 paid the highest average fare by a massive margin — £84.15 on average, compared to £20.66 for Class 2 and £13.68 for Class 3. Class 1 passengers paid over 6 times more than Class 3 passengers on average. Even the median fares tell the same story: £60.29 for Class 1, £14.25 for Class 2, and £8.05 for Class 3. The maximum fare paid (£512.33 seen in Q4) also belonged to a 1st class passenger. This confirms that fare and class are strongly linked (correlation -0.549), and 1st class passengers were high-paying wealthy travelers who received premium cabins, dining, and evacuation priority in return.

```python
avg_fare = df.groupby('Pclass')['Fare'].mean()
med_fare = df.groupby('Pclass')['Fare'].median()

plt.figure(figsize=(7, 5))
colors = ['gold', 'cornflowerblue', 'tomato']
bars = plt.bar(avg_fare.index, avg_fare.values, color=colors, edgecolor='black', width=0.5)
plt.title('Average Fare by Passenger Class')
plt.xlabel('Passenger Class')
plt.ylabel('Average Fare (£)')
plt.xticks([1, 2, 3], ['Class 1', 'Class 2', 'Class 3'])
for bar, val in zip(bars, avg_fare.values):
    plt.text(bar.get_x() + bar.get_width()/2, val + 1, f'£{val:.1f}', ha='center', fontsize=11)
plt.tight_layout()
plt.show()

print("Average Fare by Class:\n", avg_fare)
print("\nMedian Fare by Class:\n", med_fare)
```

-->

Answer:
Based on the complete visual and statistical analysis of 891 Titanic passengers, here are 6 key insights:

**Insight 1 — Gender was the strongest survival factor:**
74.2% of females survived vs only 18.9% of males — a nearly 4x difference. The "women and children first" policy was clearly enforced, making Sex the single most powerful predictor of survival in this dataset.

**Insight 2 — Passenger class directly determined survival chances:**
Class 1 had a 63% survival rate, Class 2 had 47.3%, and Class 3 had only 24.2%. Wealthier passengers in upper-deck cabins had faster lifeboat access, while 3rd class passengers were physically and structurally disadvantaged during evacuation.

**Insight 4 — Age had almost no impact on fare (correlation = 0.096):**
Despite intuitive assumptions, a passenger's age did not determine what they paid. Ticket price was driven by class of travel, not age. Young and old passengers are found across all fare levels.

**Insight 5 — Fare column has significant outliers (116 records above £65.63):**
The maximum fare of £512.33 is nearly 8x the outlier threshold. These extreme fares belong to wealthy 1st class private cabin bookings and create strong right-skew in the fare distribution. Any fare-based analysis must account for or cap these outliers.

**Insight 6 — Most passengers (60.3%) traveled alone:**
537 out of 891 passengers had a family size of 0. Family size showed only a weak positive correlation with fare (0.217), meaning solo travelers and families were spread across all classes and fare levels. Traveling alone vs with family did not strongly determine price or survival.
