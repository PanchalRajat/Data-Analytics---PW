### Question 1: Define Data Quality in the context of ETL pipelines. Why is it more than just data cleaning?

-->  

Answer:  
Data Quality in ETL pipelines refers to the accuracy, completeness, consistency, reliability, and validity of data as it moves through the extraction, transformation, and loading stages. High-quality data ensures that the information is correct, usable, and suitable for analysis and decision-making.

Data quality is more than just data cleaning because it involves a broader set of processes beyond fixing errors or handling missing values. While data cleaning focuses on correcting or removing incorrect data, data quality includes activities such as validating data against rules, ensuring consistency across systems, maintaining data integrity, and monitoring data over time.

In ETL pipelines, data quality also involves applying validation checks, standardization, and governance practices to ensure that the data remains reliable throughout the entire process, not just at a single stage. Therefore, data quality is a continuous and comprehensive approach, whereas data cleaning is only one part of it.

### Question 2: Explain why poor data quality leads to misleading dashboards and incorrect decisions.

-->  

Answer:  
Poor data quality leads to misleading dashboards because dashboards visualize whatever data they receive. If the data is inaccurate, incomplete, or inconsistent, the charts and metrics will also be incorrect.

For example, missing sales records can underreport revenue, duplicate entries can inflate numbers, and inconsistent formats can break aggregations. These issues distort trends, comparisons, and KPIs shown on dashboards.

As a result, decision-makers may take actions based on wrong insights, such as overestimating performance, misallocating resources, or targeting the wrong customer segments. Therefore, poor data quality directly leads to incorrect decisions and unreliable business outcomes.

### Question 3: What is duplicate data? Explain three causes in ETL pipelines.

-->  

Answer:  
Duplicate data refers to repeated records in a dataset where the same information appears more than once. This can lead to incorrect analysis, inflated counts, and misleading insights.

One common cause of duplicate data in ETL pipelines is **multiple data sources**. When data is collected from different systems, the same record may exist in more than one source and get loaded multiple times into the target system.

Another cause is **improper joins or transformations**. Incorrect join conditions during transformation can create duplicate rows, especially when combining tables without proper keys.

A third cause is **repeated data loading or system errors**. If ETL jobs are rerun without proper checks, the same data may be inserted again, leading to duplication.

Handling duplicate data is important to ensure accurate analysis and maintain data quality in ETL pipelines.

### Question 4: Differentiate between exact, partial, and fuzzy duplicates.

-->  

Answer:  
Duplicates in data can be categorized into different types based on how similar the records are.

**Exact duplicates** are records that are completely identical in all fields. Every value in the row matches exactly.  
Example: Two records with the same name, address, and phone number.

**Partial duplicates** are records where some key fields are the same, but other fields may differ. These records represent the same entity but contain slightly different information.  
Example: Same customer name and phone number, but different address.

**Fuzzy duplicates** are records that are similar but not exactly the same due to variations such as spelling errors, abbreviations, or formatting differences.  
Example: "Rahul Mehta" and "R. Mehta" or "Rahul Mehtha".

Identifying and handling these types of duplicates is important to ensure accurate and consistent data in ETL processes.

### Question 5: Why should data validation be performed during transformation rather than after loading?

-->  

Answer:  
Data validation should be performed during the transformation stage because it allows errors to be detected and corrected before the data is loaded into the target system.

If validation is done after loading, incorrect or inconsistent data may already be stored in the database, which can affect reports, dashboards, and decision-making. Fixing such data later becomes more complex and time-consuming.

During transformation, data can be checked against rules such as data types, ranges, formats, and business constraints. Invalid records can be corrected, rejected, or flagged before loading.

This approach ensures that only clean and reliable data enters the target system, improves data quality, and reduces the risk of errors propagating into analytics and reporting.

### Question 7: Write an SQL query on Sales_Transactions to list all duplicate keys and their counts using the business key (Customer_ID + Product_ID + Txn_Date + Txn_Amount)

-->  

```sql
SELECT
    Customer_ID,
    Product_ID,
    Txn_Date,
    Txn_Amount,
    COUNT(*) AS Duplicate_Count
FROM Sales_Transactions
GROUP BY
    Customer_ID,
    Product_ID,
    Txn_Date,
    Txn_Amount
HAVING COUNT(*) > 1;
```

Answer:  
This query identifies duplicate records based on the business key composed of Customer_ID, Product_ID, Txn_Date, and Txn_Amount. It groups the data using these columns and counts how many times each combination appears. The HAVING clause filters only those combinations that occur more than once, which represent duplicate records in the dataset.

### Question 8: Enforcing Referential Integrity

-->  

```sql
SELECT DISTINCT st.Customer_ID
FROM Sales_Transactions st
LEFT JOIN Customers_Master cm
ON st.Customer_ID = cm.CustomerID
WHERE cm.CustomerID IS NULL;
```

Answer:  
This query identifies Customer_ID values in the Sales_Transactions table that do not exist in the Customers_Master table, which violates referential integrity.

A LEFT JOIN is used to match records from Sales_Transactions with Customers_Master. If a Customer_ID from Sales_Transactions does not have a corresponding match in Customers_Master, it will result in NULL values for the joined columns.

The WHERE clause filters these unmatched records, returning only those Customer_IDs that are invalid. This helps ensure that all transaction records are linked to valid customers in the master table.
