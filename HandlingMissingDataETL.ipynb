### SECTION A – THEORETICAL QUESTIONS

### Q1. What are the most common reasons for missing data in ETL pipelines?

-->  

Answer:  
Missing data in ETL pipelines can occur for several reasons during data collection, storage, or transfer between systems.

One common reason is **data entry errors**, where users may leave certain fields blank or enter incomplete information. Another reason is **system integration issues**, where data from different source systems may not match perfectly, causing some fields to remain empty.

Missing data can also occur due to **system failures or data transfer problems**, where records are partially loaded during extraction or transformation. Additionally, **changes in database structure** or differences between source and target schemas may cause some data fields to be missing.

Sometimes data is intentionally missing because certain information was **not applicable or not collected** at the time of data entry.

Understanding the causes of missing data helps ETL developers choose the correct method to handle it during the transformation stage.

### Q2. Why is blindly deleting rows with missing values considered a bad practice in ETL?

-->  

Answer:  
Blindly deleting rows with missing values is considered a bad practice in ETL because it can lead to **loss of important data** and may affect the accuracy of analysis. If many records contain missing values, removing them can significantly reduce the dataset size and remove useful information.

Another issue is **biased results**. If the missing data is related to certain patterns or groups, deleting those rows may distort the data distribution and lead to incorrect conclusions during analysis.

In some cases, only a few columns may have missing values while the rest of the data in the row is still useful. Deleting the entire row would waste valuable information that could still be used.

Instead of blindly deleting rows, ETL developers should carefully analyze the data and use appropriate techniques such as **imputation, replacement with statistical values, or data transformation** to handle missing values effectively.

### Q3. Explain the difference between Listwise deletion and Column deletion. Also mention one scenario where each is appropriate.

-->  

Answer:  
**Listwise deletion** is a method where entire rows are removed from a dataset if they contain missing values in any column. This approach keeps only complete records in the dataset.  
**Scenario:** Listwise deletion is appropriate when only a small number of rows contain missing values. For example, if a student dataset has only 2–3 rows with missing marks out of 1000 records, those rows can be safely removed without affecting the analysis.

**Column deletion** means removing an entire column from the dataset when most of its values are missing or the column is not important for analysis.  
**Scenario:** Column deletion is appropriate when a column contains too many missing values. For example, if a “Secondary Phone Number” column has missing values for 90% of the records, it may be better to remove the entire column rather than trying to fill it.

### Q4. Why is median imputation preferred over mean imputation for skewed data such as income?

-->  

Answer:  
Median imputation is preferred over mean imputation for skewed data because the **mean is highly affected by extreme values (outliers)**, while the **median is more robust and stable**.

In datasets such as income, a small number of very high values can significantly increase the mean. If the mean is used to fill missing values, it may produce unrealistic estimates that do not represent the majority of the data.

The median represents the **middle value of the dataset**, so it is less influenced by extreme values. Because of this, median imputation provides a more accurate and reliable estimate when the data distribution is skewed.

For example, if most incomes are between ₹20,000 and ₹50,000 but a few values are above ₹5,00,000, the mean will become much higher than typical incomes. In such cases, using the median gives a better representation of the typical income in the dataset.

### Q5. What is forward fill and in what type of dataset is it most useful?

-->  

Answer:  
Forward fill is a data imputation technique used to handle missing values by replacing them with the **most recent previous valid value** in the dataset.

This method assumes that the previous value remains valid until a new value appears. Instead of leaving the value missing, the last known value is carried forward to fill the gap.

Forward fill is most useful in **time-series datasets**, where data is recorded over time, such as stock prices, sensor readings, weather data, or daily sales records. In these types of datasets, the previous value is often a reasonable estimate for the missing one until a new measurement is recorded.

For example, if a sensor records temperature every hour and one value is missing, the system may fill the missing value with the previous recorded temperature using forward fill.

### Q6. Why should flagging missing values be done before imputation in an ETL workflow?

-->  

Answer:  
Flagging missing values before imputation is important because it helps preserve information about which data points were originally missing. Once missing values are replaced through imputation, it becomes difficult to identify which values were originally absent.

By creating a **missing-value flag column** (for example, `Age_missing = 1` if Age was missing), analysts and data scientists can track where imputation has occurred. This is useful for auditing the ETL process and understanding how missing data might affect analysis results.

Flagging also helps machine learning models or analytical systems detect patterns related to missing data. In some cases, the fact that data was missing can itself carry meaningful information.

Therefore, flagging missing values before imputation improves **data transparency, traceability, and analytical reliability** in an ETL workflow.

### Q7. Consider a scenario where income is missing for many customers. How can this missingness itself provide business insights?

-->  

Answer:  
Missing income data can sometimes reveal useful business insights rather than just being treated as a data problem. If many customers have missing income values, it may indicate certain patterns in customer behavior or data collection processes.

For example, customers with higher incomes may choose not to disclose their financial details due to privacy concerns. On the other hand, missing income data may also occur more frequently among new customers or customers from specific regions where income information is not collected.

Analyzing the pattern of missing income values can help businesses identify gaps in their data collection process or understand customer segments that are less likely to share personal financial information. This insight can help improve customer surveys, data collection strategies, and targeted marketing approaches.
