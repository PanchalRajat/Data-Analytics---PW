### Question 1: Define Data Transformation in ETL and explain why it is important.

-->  

Answer:  
Data transformation is the process of converting raw data extracted from different sources into a clean, structured, and usable format before loading it into a target system such as a data warehouse.

In the ETL pipeline, transformation is the second step after data extraction. During this stage, the data is cleaned, filtered, standardized, and organized to ensure consistency and accuracy. This may include operations such as removing duplicates, correcting data formats, applying calculations, merging datasets, or converting data types.

Data transformation is important because data collected from different systems often has different formats, structures, or quality levels. Without transformation, the data may be inconsistent or difficult to analyze. By transforming the data, organizations ensure that the information is reliable, standardized, and ready for reporting, analytics, and decision-making.

-->  

Answer:  
Data cleaning is an important step in the transformation stage of ETL. It ensures that the data is accurate, consistent, and ready for analysis. Some common data cleaning activities include:

### Question 3: What is the difference between Normalization and Standardization?

-->  

Answer:  
Normalization and standardization are data transformation techniques used to adjust values so that data becomes easier to analyze and compare.

**Normalization** is the process of scaling data values to a fixed range, usually between 0 and 1. It is commonly used when different variables have different ranges and we want them to be comparable.

**Standardization**, on the other hand, transforms data so that it has a mean of 0 and a standard deviation of 1. It is used when data needs to follow a standard normal distribution.

In simple terms, normalization rescales data to a specific range, while standardization adjusts data based on statistical properties such as mean and standard deviation.

### Question 4: A dataset has missing values in the “Age” column. Suggest two techniques to handle this and explain when they should be used.

-->  

Answer:  
Missing values in the **Age** column can affect data analysis, so they must be handled during the data transformation stage. Two common techniques to handle missing values are:

1. **Mean or Median Imputation** – The missing age values can be replaced with the average (mean) or median age of the dataset. This technique is useful when the number of missing values is small and the data distribution is fairly consistent.

2. **Removing Records with Missing Values** – Rows that contain missing age values can be removed from the dataset. This method is suitable when the number of missing records is very small and removing them will not significantly affect the dataset or analysis results.

Choosing the right technique depends on how much data is missing and how important the column is for the analysis.

### Question 5: Convert the following inconsistent “Gender” entries into a standardized format (“Male”, “Female”): ["M", "male", "F", "Female", "MALE", "f"]

-->  

Answer:  
During data transformation, inconsistent values must be standardized so that the dataset becomes consistent and easier to analyze.

The values can be standardized as follows:

- "M", "male", "MALE" → **Male**  
- "F", "Female", "f" → **Female**

After transformation, the cleaned values would be:

["Male", "Male", "Female", "Female", "Male", "Female"]

This can be done by first converting all values to lowercase and then mapping them to a standard format.

Example code (Python):

```python
genders = ["M", "male", "F", "Female", "MALE", "f"]

standardized = []
for g in genders:
    g = g.lower()
    if g in ["m", "male"]:
        standardized.append("Male")
    elif g in ["f", "female"]:
        standardized.append("Female")

print(standardized)
```

This code converts each value to lowercase and then maps it to either "Male" or "Female", ensuring consistent gender values in the dataset.

### Question 6: What is One-Hot Encoding? Give an example with the categories: “Red, Blue, Green”.

-->  

Answer:  
One-Hot Encoding is a data transformation technique used to convert categorical data into a numerical format so that it can be used in machine learning models or data analysis. Instead of representing categories with a single column, One-Hot Encoding creates separate binary columns for each category.

Each category is represented with a value of **1** (present) or **0** (not present). This prevents algorithms from assuming any numerical order between categories.

For example, if the categories are **Red, Blue, Green**, the One-Hot Encoded table would look like this:

| Color | Red | Blue | Green |
|------|-----|------|-------|
| Red  | 1   | 0    | 0     |
| Blue | 0   | 1    | 0     |
| Green| 0   | 0    | 1     |

Example code (Python):

```python
import pandas as pd

colors = ["Red", "Blue", "Green"]

df = pd.DataFrame(colors, columns=["Color"])

encoded = pd.get_dummies(df, columns=["Color"])

print(encoded)
```

This converts the categorical **Color** column into three binary columns: **Red**, **Blue**, and **Green**.

### Question 7: Explain the difference between Data Integration and Data Mapping in ETL.

-->  

Answer:  
Data Integration and Data Mapping are both important steps in the ETL process, but they serve different purposes.

**Data Integration** is the process of combining data from multiple sources into a single unified dataset. The goal is to bring together data from different systems, formats, or databases so that it can be analyzed as one consistent dataset. For example, an organization may integrate customer data from a CRM system and transaction data from a sales database into a data warehouse.

**Data Mapping**, on the other hand, is the process of defining how fields from the source system correspond to fields in the target system. It specifies which source column should be transferred to which destination column and how the data should be transformed if needed.

In simple terms, **data integration focuses on merging data from different sources**, while **data mapping focuses on defining the relationships between source and target fields during the transformation process**.

### Question 8: Explain why Z-score Standardization is preferred over Min-Max Scaling when outliers exist.

-->  

Answer:  
Z-score Standardization is often preferred over Min-Max Scaling when a dataset contains outliers because it is less affected by extreme values.

**Min-Max Scaling** rescales data into a fixed range, usually between 0 and 1. It uses the minimum and maximum values of the dataset. When outliers are present, the minimum or maximum values become very large or very small, which compresses the rest of the data into a very narrow range and reduces its usefulness for analysis.

**Z-score Standardization**, on the other hand, transforms the data based on the mean and standard deviation of the dataset. Instead of relying on the minimum and maximum values, it measures how far each data point is from the mean in terms of standard deviations.

Because of this, Z-score Standardization handles outliers better and maintains the relative distribution of the data. This makes it more suitable for datasets where extreme values are present.
