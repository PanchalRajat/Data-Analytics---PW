### Task 1: Category Performance Analysis

Business Need:
Management wants to know which product categories generate the highest revenue.

Task:
Create a visualization to compare total Sales by Category and identify the top-performing category.

-->

Answer:

This task is about comparing total sales across categories to identify which category generates the highest revenue.

Steps:

1. Drag "Category" to the Rows shelf to display different product categories
2. Drag "Sales" to the Columns shelf to represent total revenue
3. Ensure aggregation is set to SUM (Sum of Sales)
4. A bar chart will be created automatically to compare categories
5. Click on Sort Descending to arrange categories from highest to lowest sales
6. Drag "Sales" to Label in the Marks card to display values on each bar
7. The category with the highest bar represents the top-performing category

### Q9. Chart Selection Justification

-->

Answer:

This task is about explaining why different types of charts were chosen for different business questions to ensure clear and effective data understanding.

Steps:

1. Bar Chart (Sales by Category / Region):
   - Used for comparison between different groups
   - Helps easily identify highest and lowest performers

2. Line Chart (Monthly Sales Trend):
   - Used to show trends over time
   - Helps identify growth, peaks, and slow periods

3. Pie Chart (Sales by Region Contribution):
   - Used to show proportion or percentage contribution
   - Helps understand share of each region in total sales

4. Scatter Plot (Sales vs Profit Analysis):
   - Used to show relationship between two variables
   - Helps identify loss-making and high-performing products

5. Heat Map (Category vs Region):
   - Used for multi-dimensional comparison
   - Helps quickly identify strong and weak areas using color intensity

6. Bubble Chart / Size Encoding (Product Prioritization):
   - Used to represent multiple variables (Sales, Profit, Quantity)
   - Helps identify priority products visually

7. Running Total Line Chart (Cumulative Growth):
   - Used to show overall growth over time
   - Helps evaluate consistency of business performance

8. Dashboard Combination:
   - Combines multiple charts for a complete view
   - Enables interactive filtering and better decision-making
