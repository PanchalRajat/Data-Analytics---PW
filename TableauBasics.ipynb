### Question 3: Create a Bar Chart showing Total Sales by Category.

Explain:

Which field goes to Rows  
Which field goes to Columns  
Which aggregation is applied to Sales  

-->  

Answer:  

Steps:

1. Open Tableau and load **Sample – Superstore** dataset.  
2. Drag **Category → Rows shelf**.  
3. Drag **Sales → Columns shelf**.  
4. Tableau will automatically create a **Bar Chart**.  
5. Ensure Sales aggregation is set to **SUM**.  

---

**Explanation:**

- **Rows:** Category → used to group data (Furniture, Technology, etc.)  
- **Columns:** Sales → used to measure total sales  
- **Aggregation:** SUM(Sales) → Tableau automatically sums sales values  

This creates a bar chart showing **total sales for each category**.

### Question 8: What is the difference between a Worksheet and a Dashboard in Tableau?  
Why is a dashboard more useful for decision-makers?

-->  

Answer:  

In Tableau, a **Worksheet** and a **Dashboard** serve different purposes:

- **Worksheet**  
  - Contains a **single chart or visualization**  
  - Used for detailed analysis of one view  
  - Example: Sales by Category chart  

- **Dashboard**  
  - Combines **multiple worksheets** in one screen  
  - Shows complete business insights in one place  
  - Supports interaction (filters, actions)

---

**Why dashboards are more useful?**

- Provide **complete view of data at once**  
- Help in **quick decision-making**  
- Allow **interactive analysis** (filters, drill-down)  
- Save time by avoiding switching between charts  

---

Steps:

1. Create multiple worksheets  
2. Click **New Dashboard**  
3. Drag worksheets into dashboard  
4. Add filters/slicers for interaction  

Dashboards help decision-makers see the **big picture quickly and clearly**.

### Question 9: What is Tableau Prep, and why is it important in real-world data projects?  
Then explain how you would:

Remove null values  
Remove duplicate records  
Change incorrect data types  

-->  

Answer:  

**Tableau Prep** is a data preparation tool used to **clean, transform, and organize data** before analysis in Tableau.

It is important because:
- Ensures **clean and accurate data**
- Improves **data quality**
- Saves time during analysis
- Handles messy real-world datasets

---

### Steps to Remove Null Values:

1. Open **Tableau Prep** and load dataset  
2. Go to the data grid  
3. Click on the column with null values  
4. Use **Filter → Exclude Null values**  
   (or manually uncheck null)  

---

### Steps to Remove Duplicate Records:

1. Add a **Clean Step**  
2. Select all relevant columns  
3. Use **Remove Duplicates** option  
   (or group + aggregate to remove duplicates)  

---

### Steps to Change Incorrect Data Types:

1. Click on the column  
2. Check current data type (icon)  
3. Change it to correct type (String, Number, Date)  
4. Validate values after conversion  

---

These steps ensure the dataset is **clean, consistent, and ready for analysis**.
