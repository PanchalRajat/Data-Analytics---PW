### Question 1

What is a Calculated Field and why is it important in business dashboards?

-->

Answer:

This question is about understanding what a calculated field is and how it helps in analyzing data in dashboards.

Steps:

1. A Calculated Field is a new field created using existing data with formulas
2. It allows users to perform custom calculations like profit ratio, discounts, or conditions
3. It helps in transforming raw data into meaningful insights
4. It enables dynamic analysis without changing the original dataset
5. It is used to create KPIs (Key Performance Indicators)
6. It improves decision-making by showing calculated metrics directly in visuals

Code:
```
Example:
Profit Ratio = SUM([Profit]) / SUM([Sales])
```

### Question 2

Why can’t parameters change visuals directly in Tableau?

-->

Answer:

This question is about understanding the role of parameters and why they cannot directly control visuals.

Steps:

1. Parameters are independent variables and not directly connected to data
2. They do not automatically affect charts or fields
3. They only store a value selected by the user
4. To use parameters, they must be linked with calculated fields
5. Calculated fields use parameter values to control logic
6. Then those calculated fields are used in visuals

Code:

```
Example:
IF [Select Metric] = "Sales" THEN SUM([Sales])
ELSE SUM([Profit])
END
```

### Question 3

When should EXCLUDE LOD be used?

-->

Answer:

This question is about understanding when to use EXCLUDE Level of Detail (LOD) expressions.

Steps:

1. EXCLUDE LOD is used when you want to remove a dimension from calculation
2. It helps compare overall values without specific breakdown
3. Useful for calculating totals ignoring a dimension
4. Used when current view has more detail but you want higher-level summary
5. Helps in comparing part-to-whole relationships

Code:

Example:
```
{ EXCLUDE [Region] : SUM([Sales]) }
```

### Question 4

Difference between FIXED and INCLUDE LOD expressions?

-->

Answer:

This question is about understanding the difference between FIXED and INCLUDE LOD expressions in Tableau.

Steps:

1. FIXED LOD:
   - Ignores the view level of detail
   - Calculates based on specified dimensions only
   - Does not change when you add/remove dimensions in view
   - Used for consistent calculations

2. INCLUDE LOD:
   - Adds extra dimensions to the current view level
   - Works with existing visualization level
   - Changes when view changes
   - Used for more detailed calculations

3. Key Difference:
   - FIXED = Independent of view
   - INCLUDE = Dependent on view

Code:
```
FIXED Example:
{ FIXED [Region] : SUM([Sales]) }

INCLUDE Example:
{ INCLUDE [Sub-Category] : SUM([Sales]) }
```

### Task 1: Delivery Performance Classification (Calculated Field)

Business Need:
Operations wants to track delivery efficiency.

Task:
Classify orders as Fast or Delayed.

Condition:
Delivery ≤ 30 min → Fast  
Else → Delayed

-->

Answer:

This task is about creating a calculated field to classify each order based on delivery time into Fast or Delayed categories and visualizing it using a single chart.

Steps:

1. Go to Data pane → Click on "Create Calculated Field"
2. Name the field as "Delivery Status"
3. Write logic using Delivery_Time_Min to classify orders
4. Click OK to create the field
5. Drag "Delivery Status" to Rows
6. Drag "Order_ID" to Columns and change it to COUNT (Count of Orders)
7. Drag "Delivery Status" to Color to differentiate Fast and Delayed
8. Drag "Order_ID" to Label to show number of orders
9. Analyze number of Fast vs Delayed deliveries

Code:

```
IF [Delivery_Time_Min] <= 30 THEN "Fast"
ELSE "Delayed"
END
```

### Task 3: Top N Products Analysis (Parameter)

Business Need:
The product team wants to focus on top-performing products.

Task:
Create a Top N Products dynamic analysis.

-->

Answer:

This task is about creating a dynamic Top N analysis using a parameter so users can choose how many top products to display in a single chart.

Steps:

1. Go to Data pane → Click on "Create Parameter"
2. Name it "Top N"
3. Set Data Type to Integer
4. Set Allowable values (Range or List, e.g., 5, 10, 15)
5. Click OK and Show Parameter on screen
6. Create a Calculated Field → name it "Top N Filter"
7. Use RANK function based on Sales
8. Click OK to create the field
9. Drag "Product_Name" to Rows
10. Drag "Sales" to Columns
11. Sort Sales in Descending order
12. Drag "Top N Filter" to Filters → select True
13. Select Bar Chart (single chart only)
14. Drag "Sales" to Label to show values
15. Rename sheet to "Top N Products"
16. Use parameter to dynamically change number of top products

Code:

Top N Filter:

```
RANK_DENSE(SUM([Sales]), 'desc') <= [Top N]
```
